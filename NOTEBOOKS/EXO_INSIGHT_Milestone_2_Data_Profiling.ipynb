## 2. Load the original NASA dataset

NASA's export may include metadata lines beginning with `#`. The `comment="#"` option tells pandas to ignore those lines when reading the table. This does not edit the source CSV.

````python
import pandas as pd
import numpy as np
from pathlib import Path
from IPython.display import display

PROJECT_DIR = Path(r"S:\Aivancity\PGE 3\Coding for AI\EXO INSIGHTS")

# Dataset paths
RAW_FILE = PROJECT_DIR / "DATA" / "raw" / "PSCompPars.csv"
PROCESSED_DIR = PROJECT_DIR / "DATA" / "processed"

# Create processed folder if it doesn't exist
PROCESSED_DIR.mkdir(parents=True, exist_ok=True)

# Verify dataset location
print("Project directory:", PROJECT_DIR)
print("Dataset location:", RAW_FILE)
print("Dataset exists:", RAW_FILE.exists())

# Load original NASA dataset
if not RAW_FILE.exists():
    raise FileNotFoundError(
        f"NASA dataset not found at: {RAW_FILE}"
    )

df = pd.read_csv(
    RAW_FILE,
    comment="#",
    low_memory=False
)

# Initial dataset information
print("\nNASA dataset loaded successfully!")
print("=" * 45)
print("Number of records:", df.shape[0])
print("Number of columns:", df.shape[1])
print("Dataset size:", round(
    RAW_FILE.stat().st_size / (1024 * 1024), 2
), "MB")

# Display first 10 records
display(df.head(10))

print("\nDataset loading completed successfully!")
````

## 4. Inspect column names and data types

````python
column_report = pd.DataFrame({
    "Column": df.columns,
    "Data_Type": [str(dtype) for dtype in df.dtypes],
    "Non_Null_Count": df.notna().sum().values,
    "Missing_Count": df.isna().sum().values
})

display(column_report)
print("Total columns:", len(df.columns))
````

## 5. Missing-value analysis

This report shows how many values are missing in each column and what percentage of the dataset that represents. Missing values are not automatically errors; some planetary properties have not been measured for every planet.

````python
missing_report = pd.DataFrame({
    "Missing_Count": df.isna().sum(),
    "Missing_Percentage": (df.isna().mean() * 100).round(2),
    "Available_Count": df.notna().sum()
}).sort_values("Missing_Percentage", ascending=False)

display(missing_report)

missing_report.to_csv(
    PROCESSED_DIR / "milestone2_missing_values.csv",
    index=True
)
print("Saved: data/processed/milestone2_missing_values.csv")
````

## 6. Duplicate-record checks

We check both:
- Repeated planet names (`pl_name`).
- Entire rows that are exact duplicates.

A repeated planet name is not automatically safe to delete. We will inspect examples and decide how to handle them later.

````python
if "pl_name" in df.columns:
    duplicate_name_mask = df["pl_name"].duplicated(keep=False)
    duplicate_name_rows = df.loc[duplicate_name_mask].sort_values("pl_name")
    duplicate_name_count = df["pl_name"].duplicated().sum()

    print("Rows belonging to repeated planet names:", len(duplicate_name_rows))
    print("Repeated name occurrences after first appearance:", duplicate_name_count)
    display(duplicate_name_rows.head(30))
else:
    duplicate_name_rows = pd.DataFrame()
    print("Column pl_name is not present.")
````

````python
exact_duplicate_count = int(df.duplicated().sum())
print("Exact duplicate rows (after first occurrence):", exact_duplicate_count)
````

## 7. Check required research columns

These are the fields we planned to use. `st_lum` is included as optional because it was not present in the earlier inspection. This cell checks the actual file rather than assuming availability.

````python
required_columns = [
    "pl_name",
    "hostname",
    "pl_rade",
    "pl_bmasse",
    "pl_bmassprov",
    "pl_eqt",
    "pl_insol",
    "pl_orbper",
    "pl_orbsmax",
    "st_teff",
    "st_rad",
    "st_mass",
    "st_lum",
    "sy_dist",
    "disc_year",
    "discoverymethod"
]

column_availability = pd.DataFrame({
    "Column": required_columns,
    "Present": [col in df.columns for col in required_columns]
})
display(column_availability)

available_required_columns = [
    col for col in required_columns if col in df.columns
]
missing_required_columns = [
    col for col in required_columns if col not in df.columns
]

print("Available planned columns:", len(available_required_columns))
print("Unavailable planned columns:", missing_required_columns)
````

## 8. Numerical summaries and unusual values

This is an initial screening, not a declaration that any value is wrong. Very large or small values may be scientifically valid. We will review suspicious values in context before defining validation rules.

````python
numeric_summary = df.describe(include=[np.number]).T
display(numeric_summary)

numeric_summary.to_csv(
    PROCESSED_DIR / "milestone2_numeric_summary.csv"
)
print("Saved: data/processed/milestone2_numeric_summary.csv")
````

````python
# Check for negative values in selected fields where negative values
# would generally require investigation.
nonnegative_fields = [
    "pl_rade", "pl_bmasse", "pl_eqt", "pl_insol",
    "pl_orbper", "pl_orbsmax", "st_teff", "st_rad",
    "st_mass", "sy_dist"
]

negative_value_report = []
for col in nonnegative_fields:
    if col in df.columns:
        values = pd.to_numeric(df[col], errors="coerce")
        negative_value_report.append({
            "Column": col,
            "Negative_Count": int((values < 0).sum())
        })

negative_value_report = pd.DataFrame(negative_value_report)
display(negative_value_report)
````

## 9. Discovery-year coverage

The discovery year is not the same as the date the dataset was downloaded. This section reports the earliest and latest discovery years represented in this file.

````python
if "disc_year" in df.columns:
    discovery_year = pd.to_numeric(df["disc_year"], errors="coerce")
    print("Earliest discovery year:", discovery_year.min())
    print("Latest discovery year:", discovery_year.max())
    print("Records with missing discovery year:", discovery_year.isna().sum())
    display(discovery_year.value_counts().sort_index().tail(15))
else:
    print("disc_year is not available.")
````

## 10. Mass provenance inspection

NASA's `pl_bmassprov` field describes the provenance/type of the reported mass estimate. Inspect its categories before deciding whether to treat all mass estimates alike.

````python
if "pl_bmassprov" in df.columns:
    mass_provenance_report = (
        df["pl_bmassprov"]
        .fillna("MISSING")
        .value_counts(dropna=False)
        .rename_axis("Mass_Provenance")
        .reset_index(name="Record_Count")
    )
    display(mass_provenance_report)
    mass_provenance_report.to_csv(
        PROCESSED_DIR / "milestone2_mass_provenance.csv",
        index=False
    )
else:
    print("pl_bmassprov is not available.")
````

## 11. Save a compact profiling summary

This creates a reproducible summary file. It does not create a cleaned dataset or alter the original NASA file.

````python
summary = {
    "source_file": RAW_FILE.name,
    "row_count": int(df.shape[0]),
    "column_count": int(df.shape[1]),
    "exact_duplicate_rows": int(df.duplicated().sum()),
    "repeated_planet_name_rows": (
        int(df["pl_name"].duplicated(keep=False).sum())
        if "pl_name" in df.columns else None
    ),
    "available_planned_columns": len(available_required_columns),
    "unavailable_planned_columns": ", ".join(missing_required_columns)
}

if "disc_year" in df.columns:
    years = pd.to_numeric(df["disc_year"], errors="coerce")
    summary["earliest_discovery_year"] = (
        int(years.min()) if years.notna().any() else None
    )
    summary["latest_discovery_year"] = (
        int(years.max()) if years.notna().any() else None
    )

summary_df = pd.DataFrame([summary])
display(summary_df)

summary_df.to_csv(
    PROCESSED_DIR / "milestone2_profile_summary.csv",
    index=False
)
print("Saved profiling reports in:", PROCESSED_DIR)
````
